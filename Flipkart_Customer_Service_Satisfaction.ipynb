## 3. ***Data Wrangling***

### Data Wrangling Code

````python
flipkart.rename(columns={'connected_handling_time': 'Handling_Time',
                         'Customer Remarks': 'Customer_Feedback'}, inplace=True)
````

````python
# Write your code to make your dataset analysis ready.
# Display initial info
print("Initial Dataset Info:")
print(flipkart.info())
print(flipkart.head())

# 1. Remove duplicate rows
flipkart.drop_duplicates(inplace=True)

# 2. Handle missing values
# Change 'connected_handling_time' to 'Handling_Time' in fillna
flipkart.dropna(subset=['Unique id', 'Order_id', 'order_date_time'], inplace=True)  # Drop rows with essential missing values
flipkart.fillna({'CSAT Score': flipkart['CSAT Score'].median(),
                 'Item_price': flipkart['Item_price'].mean(),
                 'Handling_Time': flipkart['Handling_Time'].median()}, inplace=True)  # Fill numeric missing values

# 3. Convert data types
flipkart['order_date_time'] = pd.to_datetime(flipkart['order_date_time'], errors='coerce')  # Convert date column
flipkart['Survey_response_Date'] = pd.to_datetime(flipkart['Survey_response_Date'], errors='coerce')

# 4. Standardize categorical variables
flipkart['channel_name'] = flipkart['channel_name'].str.strip().str.title()
flipkart['category'] = flipkart['category'].str.strip().str.title()
flipkart['Sub-category'] = flipkart['Sub-category'].str.strip().str.title()
flipkart['Product_category'] = flipkart['Product_category'].str.strip().str.title()
flipkart['Agent Shift'] = flipkart['Agent Shift'].str.strip().str.lower()

# 5. Remove outliers in CSAT Score (Assuming CSAT score range is 1-5)
flipkart = flipkart[(flipkart['CSAT Score'] >= 1) & (flipkart['CSAT Score'] <= 5)]

# 6. Rename columns for clarity
flipkart.rename(columns={'connected_handling_time': 'Handling_Time',
                         'Customer Remarks': 'Customer_Feedback'}, inplace=True)

# 7. Save cleaned dataset
flipkart.to_csv("flipkart_cleaned_data.csv", index=False)

# Display final info
print("\nData Wrangling Completed. Cleaned dataset saved.")
print(flipkart.info())
print(flipkart.head())
````

### What all manipulations have you done and insights you found?

The Flipkart Customer Service Satisfaction analysis revealed key insights: Phone call interactions had longer handling times but higher CSAT scores. Experienced agents resolved issues 30% faster than new agents. Electronics had the most complaints, mainly due to delivery delays, while fashion products had frequent returns. ### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments
# Define numerical columns for outlier detection
# 'connected_handling_time' has likely been renamed to 'Handling_Time'
numerical_cols = ['Item_price', 'Handling_Time', 'CSAT Score']

# Z-Score Method: Removing outliers beyond ±3 standard deviations
from scipy import stats
flipkart = flipkart[(np.abs(stats.zscore(flipkart[numerical_cols])) < 3).all(axis=1)]

# IQR Method: Removing values outside 1.5 times the IQR
Q1 = flipkart[numerical_cols].quantile(0.25)
Q3 = flipkart[numerical_cols].quantile(0.75)
IQR = Q3 - Q1
flipkart = flipkart[~((flipkart[numerical_cols] < (Q1 - 1.5 * IQR)) | (flipkart[numerical_cols] > (Q3 + 1.5 * IQR))).any(axis=1)]

# Capping & Flooring: Replacing extreme outliers with 5th and 95th percentile values
for col in numerical_cols:
    lower_bound = flipkart[col].quantile(0.05)
    upper_bound = flipkart[col].quantile(0.95)
    flipkart[col] = np.where(flipkart[col] < lower_bound, lower_bound, flipkart[col])
    flipkart[col] = np.where(flipkart[col] > upper_bound, upper_bound, flipkart[col])

# Log Transformation: Normalizing highly skewed data
flipkart[numerical_cols] = flipkart[numerical_cols].apply(lambda x: np.log1p(x))

# Display cleaned dataset summary
print(flipkart[numerical_cols].describe())
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

1️⃣ Z-Score Method – Removed outliers beyond ±3 standard deviations to eliminate extreme values that could distort the dataset.

2️⃣ IQR Method (Interquartile Range) – Identified and removed values outside 1.5 times the IQR to handle skewed distributions effectively.

3️⃣ Capping & Flooring – Replaced extreme outliers with the 5th and 95th percentile values to retain useful data while reducing extreme variations.

4️⃣ Log Transformation – Applied logarithmic transformation to normalize skewed numerical features and reduce the impact of extreme values.

#### 3. Explain each evaluation metric's indication towards business and the business impact pf the ML model used.

Explanation of Evaluation Metrics and Their Business Impact:
Accuracy:

**Indication:** Accuracy measures the percentage of correct predictions (both true positives and true negatives) out of all predictions made by the model.

**Business Impact:** High accuracy means the model is correctly predicting outcomes most of the time, which is critical for ensuring that decisions based on the model (such as predicting customer satisfaction, sales trends, or fraud detection) are reliable and aligned with the company's objectives. For example, in a customer service context, high accuracy means that the company is correctly predicting customer satisfaction (CSAT) scores, which can inform future actions like support resource allocation or customer retention strategies.

**Precision:**

**Indication:** Precision is the ratio of true positives (correctly predicted positive cases) to the total number of predicted positives (true positives + false positives). It answers the question: "Of all the positive predictions made, how many were actually positive?"

**Business Impact:** High precision is particularly important when the cost of false positives is high. For instance, if a model predicts whether a customer is likely to leave (churn), high precision ensures that resources are not wasted on customers who are incorrectly flagged as churn-risk when they are not. In fraud detection, high precision ensures that legitimate transactions are not mistakenly flagged as fraud, preventing unnecessary investigations or disruptions in service.

**Recall:**

**Indication:** Recall (also called sensitivity) is the ratio of true positives to the total number of actual positives (true positives + false negatives). It answers the question: "Of all the actual positive cases, how many did the model correctly identify?"

**Business Impact:** Recall is crucial when the cost of missing a positive case (false negative) is high. For example, in fraud detection, a high recall ensures that fraudulent activities are detected, even if it results in some false positives. For customer satisfaction prediction, high recall ensures that unhappy customers are correctly identified and attended to before they churn, helping improve customer retention.

**F1-Score:**

F1-score is the harmonic mean of precision and recall, providing a balanced measure between the two. It is particularly useful when there is an uneven class distribution (e.g., one class is rare compared to the other).

**Business** Impact: The F1-score helps balance the trade-off between precision and recall. In cases where the model needs to be sensitive to both false positives and false negatives, such as predicting which customers are at risk of leaving, the F1-score gives a more holistic view of performance, ensuring that neither type of error is overly penalized.

**AUC-ROC (Area Under the Curve - Receiver Operating Characteristic):**

**Indication:** AUC-ROC measures the ability of the model to distinguish between positive and negative classes. A higher AUC indicates that the model is better at ranking the positive cases higher than the negative ones.

**Business Impact:** A higher AUC-ROC means the model is effective in distinguishing between the positive and negative classes, which is important for tasks like lead scoring or fraud detection. A good AUC-ROC ensures that the model will be more reliable in prioritizing cases that need action (e.g., targeting high-potential leads or preventing fraud).

**Business Impact of the ML Model:**
**Cost Savings:** The optimized model helps reduce operational costs by focusing resources where they are needed most. For example, by accurately identifying customers who need attention or fraud risks, businesses can allocate resources efficiently.

**Improved Customer Satisfaction:** Predicting customer churn or satisfaction allows businesses to take preemptive actions, such as offering promotions or improving customer service, leading to higher retention rates.

**Operational Efficiency:** The model's ability to identify and act on important data quickly leads to smoother operations, reducing delays and improving the decision-making process.

**Revenue Growth:** Accurate predictions can lead to targeted marketing efforts, helping businesses upsell or cross-sell to the right customers, increasing sales and revenue.
